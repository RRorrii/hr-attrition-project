```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
pd.set_option('display.max_columns', None)
```

```

df = pd.read_csv('../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv')


df.head()
```

```

print(f"Размер датасета: {df.shape[0]} строк, {df.shape[1]} столбцов\n")


df.info()
```

```

print(df['Attrition'].value_counts())
print("\nВ процентах:")
print(df['Attrition'].value_counts(normalize=True) * 100)


sns.countplot(data=df, x='OverTime', hue='Attrition', palette='Set1')
plt.title('Влияние переработок (OverTime) на увольнение')
plt.show()
```

```
sns.countplot(data=df, x='OverTime', hue='Attrition', palette='Set1')
plt.title('Влияние переработок (OverTime) на увольнение')
plt.show()
```

```
plt.figure(figsize=(12, 5))
sns.histplot(data=df, x='Age', hue='Attrition', kde=True, palette='Set1', bins=30)
plt.title('Распределение возраста уволившихся и оставшихся')
plt.show()
```

```

for col in df.columns:
    if df[col].nunique() == 1:
        print(f"Признак '{col}' имеет только одно значение: {df[col].unique()[0]}")


cols_to_drop = ['EmployeeCount', 'Over18', 'StandardHours']
df = df.drop(columns=cols_to_drop)
print("\nМусорные признаки удалены!")
```

```

df['Attrition_Num'] = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)


numeric_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(16, 12))

sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm', fmt=".2f")
plt.title('Корреляционная матрица числовых признаков')
plt.show()
```

```
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Department', hue='Attrition', palette='Set2')
plt.title('Увольнения в разрезе департаментов')
plt.xticks(rotation=45)
plt.show()
```
